```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('adult.csv')
df.head()
```

```python
df.replace('?', np.nan, inplace=True)
print(df.dtypes)
```

```python
#Task 1

#average age by workclass

avg_age_by_workclass = df.groupby('workclass')['age'].mean()
print("Average age by Worclass: \n", avg_age_by_workclass)
plt.bar(avg_age_by_workclass.index, avg_age_by_workclass.values)
plt.xlabel('Workclass')
plt.ylabel('Average Age')
plt.title('Average Age by Workclass')
plt.xticks(rotation=90)
plt.show()
```

```python
#Task 2

workclass_counts = df['workclass'].value_counts()

plt.bar(workclass_counts.index, workclass_counts.values)
plt.xlabel('Workclass')
plt.ylabel('Count')
plt.title('Workclass Counts')
plt.xticks(rotation=90)
plt.show()
```

```python
#Task 3

sns.histplot(df['age'], bins = 20)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution')
plt.show()

print("Data is Right Skewed because the tail is towards the right")
```

```python
#Task 4

sns.boxplot(df['age'])
plt.title('Boxplot of Age')
plt.show()
```

```python
#Task 5

Q1 = df['age'].quantile(0.25)
Q3 = df['age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['age'] < lower) | (df['age'] > upper)]
print("Number of Outliers: ", len(outliers))
```

```python
#Task 6

from scipy import stats
df['zscore'] = stats.zscore(df['age'])
outliers_z = df[df['zscore'].abs() > 3]
print("Number of Outliers: ", len(outliers_z))
print("IQR and Z-score flag different number of outliers because IQR method is better used for skewed data while Z-score is used in normal distribution.")
```

```python
#Task 7

sns.scatterplot(x=df['age'],y= df['capital.loss'], hue = df['workclass'])
plt.xlabel('Age')
plt.ylabel('Hours per Week')
plt.title('Hours per Week for Age')
plt.show()
```

```python
#Task 8

avg_age = df.groupby('education.num')['age'].mean()
plt.plot(avg_age)
plt.xlabel('Education Number')
plt.ylabel('Age')
plt.title('Education Number Distribution')
plt.show()

plt.fill_between(avg_age.index, avg_age.values)
plt.xlabel('Education Number')
plt.ylabel('Age')
plt.title('Education Number Distribution')
plt.show()
```

```python
#Task 9

plt.bar(workclass_counts.index, workclass_counts.values)
plt.ylim(bottom=2000)
plt.xlabel('Workclass')
plt.ylabel('Count')
plt.title('Workclass Counts')
plt.xticks(rotation=90)
plt.show()
```
