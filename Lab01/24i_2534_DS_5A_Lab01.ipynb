```python
import pandas as pd
import numpy as np
```

```python
# Task 1

df = pd.read_csv('/adult.csv')
df.head()
```

```python
print(df.isnull().sum())
df.replace('?', np.nan, inplace=True)
count = df.isnull().sum()
print(count)
```

```python
# Task 2

rows_count, columns_count = df.shape
percentage = (count * 100 / rows_count)
print(percentage.sort_values(ascending = False))
```

```python
#Task 3

df['workclass'].replace(np.nan, "Unknown", inplace=True)
df['occupation'].replace(np.nan, "Unknown", inplace=True)
df.head()

#Null values in both workclass and occupation has been replaced with "Unknown".
```

```python
#Task 4

print("Duplicated Rows: ", df.duplicated().sum())
```

```python
#Task 5

print("Unique values of Education: \n",df['education'].unique())
print("\nUnique values of Marital Status: \n", df['marital.status'].unique())
print("\nUnique values of Native Country: \n", df['native.country'].unique())

```

```python
from numpy._core.defchararray import isnumeric
# Task 6

print(df.describe(include = [np.number]))
df.describe(include = "object")
```

```python
# Task 7

df.value_counts('occupation')
print("The most common occupation is: Prof-speciality")

male_count, female_count = df.value_counts('sex')
print("Female percentage: ", female_count * 100 / (male_count + female_count))
print("Male percentage: ", male_count * 100 / (male_count + female_count))

income_less_than_50k, income_greater_than_50k = df.value_counts('income')
print("Percentage of Income >= 50k: ", income_greater_than_50k * 100 / (income_greater_than_50k + income_less_than_50k))
print("Percentage of Income <= 50k: ", income_less_than_50k * 100 / (income_greater_than_50k + income_less_than_50k))
```

```python
# Task 8

df.groupby('education')['education.num'].unique()
```

```python
# Task 9

import sqlite3

conn = sqlite3.connect(':memory:')

df.to_sql('adult_income', conn, if_exists='replace', index=False)

sql_query = "SELECT * FROM adult_income WHERE age > 30;"

df_sql = pd.read_sql_query(sql_query, conn)

print(f"Original DataFrame shape: {df.shape}")
print(f"SQL Query Result DataFrame shape (age > 30): {df_sql.shape}")

df_filtered_original = df[df['age'] > 30]
print(f"Original DataFrame filtered (age > 30) shape: {df_filtered_original.shape}")

if df_sql.shape == df_filtered_original.shape:
    print("\nShape comparison successful: The SQL extracted DataFrame matches the shape of the original DataFrame filtered by the same condition.")
else:
    print("\nShape comparison failed: Mismatch in DataFrame shapes.")

conn.close()
```
